```python
# --- installs (uncomment if running in a fresh environment) ---
# !pip install numpy matplotlib ripser persim

import numpy as np
import matplotlib.pyplot as plt
from ripser import ripser
from persim import plot_diagrams, bottleneck

# -------------------------------
# 1) Utilities: signals + embedding
# -------------------------------
def make_signals(n=2000, dt=0.01, seed=0):
    rng = np.random.default_rng(seed)
    t = np.arange(n) * dt

    # A) periodic: sine wave + small noise
    xA = np.sin(2*np.pi*1.0*t) + 0.05*rng.normal(size=n)

    # B) quasi-periodic: two incommensurate sines (torus-like in embedding)
    xB = np.sin(2*np.pi*1.0*t) + 0.7*np.sin(2*np.pi*np.sqrt(2)*t + 0.3) + 0.05*rng.normal(size=n)

    # C) regime shift: frequency + noise jump at halfway
    xC = np.empty(n)
    mid = n//2
    xC[:mid]  = np.sin(2*np.pi*1.0*t[:mid]) + 0.05*rng.normal(size=mid)
    xC[mid:]  = np.sin(2*np.pi*2.0*t[mid:] + 0.5) + 0.20*rng.normal(size=n-mid)

    return t, xA, xB, xC

def takens_embed(x, m=3, tau=8):
    """Return delay embedding in R^m with delay tau (in samples)."""
    N = len(x) - (m-1)*tau
    if N <= 0:
        raise ValueError("Time series too short for given m, tau")
    Y = np.column_stack([x[i : i+N] for i in range(0, m*tau, tau)])
    return Y

def ph_vr(point_cloud, maxdim=2):
    """Persistent homology via Vietoris–Rips (ripser)."""
    res = ripser(point_cloud, maxdim=maxdim)
    return res["dgms"]

# -------------------------------
# 2) Build the synthetic data
# -------------------------------
t, xA, xB, xC = make_signals()

# Parameters for Takens
m, tau = 3, 12           # 3D embedding (alpha would also be possible; here we use VR)
YA = takens_embed(xA, m=m, tau=tau)
YB = takens_embed(xB, m=m, tau=tau)
YC = takens_embed(xC, m=m, tau=tau)

# -------------------------------
# 3) PH on each case
# -------------------------------
dA = ph_vr(YA, maxdim=2)
dB = ph_vr(YB, maxdim=2)
dC = ph_vr(YC, maxdim=2)

# -------------------------------
# 4) Plots: embeddings + persistence diagrams
# -------------------------------
def plot_embedding_3d(Y, title):
    from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
    fig = plt.figure(figsize=(4.5,4.0))
    ax = fig.add_subplot(111, projection='3d')
    s = max(1, 8000//len(Y))  # subsample for speed if huge
    ax.scatter(Y[::s,0], Y[::s,1], Y[::s,2], s=3)
    ax.set_title(title)
    plt.show()

# embeddings
plot_embedding_3d(YA, "A) Periodic — 3D delay embedding (circle)")
plot_embedding_3d(YB, "B) Quasi-periodic — 3D delay embedding (torus-like)")
plot_embedding_3d(YC, "C) Regime shift — 3D delay embedding (mixed)")

# persistence diagrams (standard birth–death with diagonal)
fig, axes = plt.subplots(3, 2, figsize=(9, 10))
plot_diagrams(dA, ax=axes[0,0]); axes[0,0].set_title("A) Periodic — PD (H0,H1,H2)")
plot_diagrams(dB, ax=axes[1,0]); axes[1,0].set_title("B) Quasi-periodic — PD (H0,H1,H2)")
plot_diagrams(dC, ax=axes[2,0]); axes[2,0].set_title("C) Regime shift — PD (H0,H1,H2)")

# highlight H1 only (loops)
plot_diagrams([dA[1]], ax=axes[0,1]); axes[0,1].set_title("A) H1 only (loop strength)")
plot_diagrams([dB[1]], ax=axes[1,1]); axes[1,1].set_title("B) H1 only (two loops?)")
plot_diagrams([dC[1]], ax=axes[2,1]); axes[2,1].set_title("C) H1 only")
plt.tight_layout(); plt.show()

# -------------------------------
# 5) Sliding-window regime detection via bottleneck distance on H1
# -------------------------------
def sliding_windows(x, w, step):
    """Return array of windows (n_windows, w)."""
    idx = np.arange(0, len(x)-w+1, step)
    return np.array([x[i:i+w] for i in idx]), idx

# pick window params (seconds ~ w*dt)
dt = t[1]-t[0]
w = int(4.0/dt)          # 4 seconds per window
step = int(0.5/dt)       # slide every 0.5 s

Xwin, starts = sliding_windows(xC, w=w, step=step)

def h1_diagram_of_window(xwin, m=3, tau=12):
    Y = takens_embed(xwin, m=m, tau=tau)
    dgm = ripser(Y, maxdim=1)["dgms"][1]  # H1 only
    return dgm

# baseline: first window
base_dgm = h1_diagram_of_window(Xwin[0], m=m, tau=tau)

# bottleneck distance to baseline for each window
bn = []
for xw in Xwin:
    dgm = h1_diagram_of_window(xw, m=m, tau=tau)
    bn.append(bottleneck(base_dgm, dgm))
bn = np.array(bn)

# plot the time series + bottleneck distances
fig, ax = plt.subplots(2, 1, figsize=(10,6), sharex=False)
ax[0].plot(t, xC, lw=1)
ax[0].axvline(t[len(xC)//2], color='k', ls='--', alpha=0.6)
ax[0].set_title("C) Regime shift time series (vertical line = change point)")
ax[0].set_ylabel("x(t)")

centers = t[starts + w//2]
ax[1].plot(centers, bn, lw=1.5)
ax[1].axhline(np.median(bn)+2*np.std(bn), ls='--', alpha=0.6)  # quick visual threshold
ax[1].set_title("Bottleneck distance (H1 PD) to first window")
ax[1].set_ylabel("d_B"); ax[1].set_xlabel("time")
plt.tight_layout(); plt.show()
```
