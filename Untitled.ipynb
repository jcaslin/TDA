```python
import numpy as np
import matplotlib.pyplot as plt
from tadasets import torus
from ripser import ripser
import persim

# 1) Sample a 3D donut torus
#    c = major radius (center-to-tube-center), a = tube radius
X = torus(n=2000, c=2.0, a=0.7, noise=0.01, seed=42, uniform=True)  # shape (n, 3)

# 2) (Optional) quick 3D visual
fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:,0], X[:,1], X[:,2], s=2)
ax.set_box_aspect((1,1,1))
ax.set_title("tadasets.torus (c=2.0, a=0.7)")
plt.show()

# 3) Persistent homology (VR)
dgms = ripser(X)['dgms']

# 4) Plot persistence diagrams (with diagonal)
persim.plot_diagrams(dgms, show=True)
```
